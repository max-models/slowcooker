# Examples

The notebooks in this section are executed when the documentation is built. Download a notebook from its page to run it yourself.

- [Geometry from GVEC](/slowcooker/examples/gvec-geometry/): the flux-surface averages of W7-X, the TORAX geometry section, and a check on a circular tokamak.
- [Heat transport in W7-X](/slowcooker/examples/w7x-heat-transport/): constant diffusivities, the stellarator neoclassical model, a power scan.
- [Beams in W7-X](/slowcooker/examples/nbi-heating/): TORAX coupled to freeofcharge's neutral beams.
- [Pressure and geometry](/slowcooker/examples/pressure-and-geometry/): GVEC re-solved with TORAX's pressure.
- [Scripts](/slowcooker/example-scripts/): complete scripts, among them a W7-X run with everything coupled.
